Uma empresa de eficiência energética está analisando o comportamento de uma residência de baixo consumo. A equipe deseja identificar períodos de consumo elevado dos eletrodomésticos e observar quais condições de temperatura e umidade estavam presentes nesses momentos.

2.	Renomeie Appliances para Consumo_Eletrodomesticos e simplifique o nome de pelo menos três atributos ambientais.

```python
df1.rename(columns={'Appliances':'Consumo_Eletrodomesticos',
                    'T1': 'Temperatura_1',
                    'T2': 'Temperatura_2',
                    'T3': 'Temperatura_3'
                    }, inplace=True)
```

```python
df1.head()
```

3. **Determine o maior consumo de eletrodomésticos registrado na amostra.**

```python
# Maior consumo de eletrodomésticos registrado na amostra
maior_consumo = df1['Consumo_Eletrodomesticos'].max()

print(f'Maior consumo de eletrodomésticos: {maior_consumo:.2f}')
```

4. **Calcule um limiar correspondente a 70% do valor máximo e crie um DataFrame contendo somente os registros acima desse limite.**

```python
# Cálculo do limiar de 70% do maior consumo
limiar_70 = 0.70 * maior_consumo

# Seleção dos registros acima do limiar
df_consumo_alto = df1[df1['Consumo_Eletrodomesticos'] > limiar_70].copy()

print(f'Limiar de 70%: {limiar_70:.2f}')
df_consumo_alto.head()
```

5. **Conte os registros selecionados e calcule qual percentual da amostra eles representam.**

```python
# Quantidade e percentual de registros com consumo acima de 70% do máximo
quantidade_alto = df_consumo_alto.shape[0]
percentual_alto = quantidade_alto / df1.shape[0] * 100

print(f'Registros acima de 70% do máximo: {quantidade_alto}')
print(f'Percentual da amostra: {percentual_alto:.2f}%')
```

6. **Calcule a temperatura média de T1 e crie outro DataFrame contendo simultaneamente consumo acima de 70% do máximo e temperatura acima da média.**

```python
# Temperatura média de T1
media_temperatura_1 = df1['Temperatura_1'].mean()

# Aplicação simultânea dos dois critérios
df_consumo_alto_temp = df1[
    (df1['Consumo_Eletrodomesticos'] > limiar_70) &
    (df1['Temperatura_1'] > media_temperatura_1)
].copy()

print(f'Temperatura média (T1): {media_temperatura_1:.2f}')
print(f'Registros que atendem aos dois critérios: {df_consumo_alto_temp.shape[0]}')

df_consumo_alto_temp.head()
```

7. **Compare os dois DataFrames e explique o efeito da inclusão da temperatura como segundo critério.**

Ao comparar `df_consumo_alto` com `df_consumo_alto_temp`, o segundo DataFrame possui apenas os registros que atendem **simultaneamente** aos dois critérios: consumo acima de 70% do máximo e temperatura T1 acima da média. Portanto, a inclusão da temperatura funciona como um filtro adicional e tende a reduzir a quantidade de registros selecionados. O resultado permite observar quais períodos de consumo elevado também ocorreram em condições de temperatura T1 acima do comportamento médio da amostra.

```python
# Comparação entre os dois conjuntos
qtd_consumo_alto = df_consumo_alto.shape[0]
qtd_consumo_alto_temp = df_consumo_alto_temp.shape[0]

print(f'Registros apenas com consumo alto: {qtd_consumo_alto}')
print(f'Registros com consumo alto + T1 acima da média: {qtd_consumo_alto_temp}')
print(f'Redução de registros: {qtd_consumo_alto - qtd_consumo_alto_temp}')

if qtd_consumo_alto > 0:
    percentual_dos_altos = qtd_consumo_alto_temp / qtd_consumo_alto * 100
    print(f'Percentual dos registros de consumo alto que também possuem T1 acima da média: {percentual_dos_altos:.2f}%')
```

# **Dataset 2 — Steel Industry Energy Consumption (UCI)**

**Situação**

A equipe de gestão de energia de uma indústria siderúrgica deseja localizar situações de consumo elevado e verificar se esses períodos coincidem com condições de carga elevada ou valores menos favoráveis de fator de potência.

### **1. Carregue a amostra e renomeie `Usage_kWh` para `Consumo_kWh`, simplificando também os nomes dos atributos de fator de potência.

```python
# Carregando a amostra
df2 = pd.read_csv('/content/EX_02_steel_industry.csv')

# Renomeando algumas colunas
df2 = df2.rename(columns={
    'Usage_kWh': 'Consumo_kWh',
    'Lagging_Current_Power_Factor': 'Fator_Potencia_Atrasado',
    'Leading_Current_Power_Factor': 'Fator_Potencia_Adiantado'
})

df2.head()
```

### **2. Apresente uma inspeção inicial com `head()`, `shape`, `info()` e `describe()`.**

```python
print("Primeiros registros:")
display(df2.head())

print("Tamanho da amostra:")
print(df2.shape)

print("\nInformações:")
df2.info()

print("\nEstatísticas:")
display(df2.describe())
```

### **3. Determine o maior consumo registrado e calcule um limiar de 75% do consumo máximo.

```python
maior_consumo = df2['Consumo_kWh'].max()
limiar_75 = 0.75 * maior_consumo

print(f"Maior consumo: {maior_consumo:.2f} kWh")
print(f"Limiar de 75%: {limiar_75:.2f} kWh")
```

### **4. Crie um DataFrame contendo os registros acima desse limiar e calcule a quantidade e o percentual de registros.

```python
df_consumo_alto = df2[
    df2['Consumo_kWh'] > limiar_75
].copy()

quantidade_alto = df_consumo_alto.shape[0]
percentual_alto = quantidade_alto / df2.shape[0] * 100

print(f"Quantidade de registros: {quantidade_alto}")
print(f"Percentual da amostra: {percentual_alto:.2f}%")

df_consumo_alto.head()
```

### **5. Verifique quantos registros desse DataFrame pertencem à categoria `Maximum Load`.**

```python
quantidade_maximum_load = (
    df_consumo_alto['Load_Type'] == 'Maximum Load'
).sum()

print(
    f"Registros classificados como Maximum Load: "
    f"{quantidade_maximum_load}"
)

print("\nDistribuição das categorias:")
display(df_consumo_alto['Load_Type'].value_counts())
```

### **6. Escolha um dos atributos de fator de potência, observe seus valores e defina um limite coerente para representar valores mais baixos.

```python
# Vamos analisar o fator de potência atrasado
print(df2['Fator_Potencia_Atrasado'].describe())

print("\nAlguns dos menores valores:")
display(
    df2[['Fator_Potencia_Atrasado']]
    .sort_values('Fator_Potencia_Atrasado')
    .head()
)

# Limite escolhido
limite_fp = 0.90

print(f"\nLimite escolhido: {limite_fp}")
```

### **7. Crie um novo DataFrame contendo simultaneamente consumo elevado e fator de potência abaixo do limite escolhido.

```python
df_consumo_alto_fp = df2[
    (df2['Consumo_kWh'] > limiar_75) &
    (df2['Fator_Potencia_Atrasado'] < limite_fp)
].copy()

quantidade_alto_fp = df_consumo_alto_fp.shape[0]
percentual_alto_fp = quantidade_alto_fp / df2.shape[0] * 100

print(f"Quantidade de registros: {quantidade_alto_fp}")
print(f"Percentual da amostra: {percentual_alto_fp:.2f}%")

df_consumo_alto_fp.head()
```

### **8. Explique por que esse segundo conjunto de registros pode merecer maior atenção da equipe de energia.**

Esse segundo conjunto pode merecer mais atenção porque apresenta, ao mesmo tempo, consumo elevado e fator de potência abaixo do limite escolhido. Assim, são registros que juntam duas condições que podem ser interessantes para a equipe analisar com mais cuidado.

O filtro não significa que exista necessariamente um problema, mas ajuda a encontrar situações que podem ser investigadas.

```python
# Comparando os dois DataFrames
print(f"Registros com consumo alto: {quantidade_alto}")
print(f"Registros com consumo alto + fator de potência baixo: {quantidade_alto_fp}")

if quantidade_alto > 0:
    percentual_dos_altos = quantidade_alto_fp / quantidade_alto * 100
    print(
        f"Dos registros com consumo alto, {percentual_dos_altos:.2f}% "
        "também têm fator de potência abaixo de 0,90."
    )
```
